## Formulación del problema analítico

El problema de negocio se traduce en un problema de clasificación binaria supervisada.

La variable objetivo es `DiagPeriodL90D`, que indica si el diagnóstico de cáncer de mama ocurrió dentro de los primeros 90 días desde el screening.

| Valor de la variable objetivo | Interpretación |
|---|---|
| 1 | El diagnóstico ocurrió dentro de los primeros 90 días |
| 0 | El diagnóstico ocurrió después de los primeros 90 días |

Por lo tanto, el objetivo analítico consiste en predecir la probabilidad de que una paciente pertenezca a la clase 1, es decir, que el diagnóstico ocurra dentro de los primeros 90 días.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

sns.set_theme(style="whitegrid", palette="Set2")

OUTPUT_DIR = Path("/content/resultados_fase2")
OUTPUT_DIR.mkdir(exist_ok=True)

train = pd.read_csv("training.csv")
test = pd.read_csv("test.csv")
sample_submission = pd.read_csv("sample_submission.csv")

RENOMBRE_COLUMNAS = {
    "patient_id": "id_paciente",
    "DiagPeriodL90D": "diagnostico_90_dias",

    "patient_race": "raza_paciente",
    "payer_type": "tipo_seguro_pagador",
    "patient_state": "estado_paciente",
    "patient_zip3": "codigo_postal_3",
    "patient_age": "edad_paciente",
    "patient_gender": "sexo_paciente",
    "bmi": "indice_masa_corporal",

    "breast_cancer_diagnosis_code": "codigo_diagnostico_cancer_mama",
    "breast_cancer_diagnosis_desc": "descripcion_diagnostico_cancer_mama",
    "metastatic_cancer_diagnosis_code": "codigo_diagnostico_cancer_metastasico",
    "metastatic_first_novel_treatment": "primer_tratamiento_novedoso",
    "metastatic_first_novel_treatment_type": "tipo_primer_tratamiento_novedoso",

    "Region": "region_geografica",
    "Division": "division_geografica",

    "population": "poblacion_zona",
    "density": "densidad_poblacional",
    "age_median": "edad_mediana_zona",

    "income_household_median": "ingreso_mediano_hogar",
    "income_individual_median": "ingreso_mediano_individual",
    "home_ownership": "porcentaje_vivienda_propia",
    "home_value": "valor_mediano_vivienda",
    "rent_median": "renta_mediana",
    "rent_burden": "carga_renta_ingreso",

    "education_less_highschool": "educacion_menor_secundaria",
    "education_highschool": "educacion_secundaria",
    "education_some_college": "educacion_superior_incompleta",
    "education_bachelors": "educacion_universitaria",
    "education_graduate": "educacion_posgrado",
    "education_college_or_above": "educacion_universitaria_o_mas",
    "education_stem_degree": "educacion_stem",

    "labor_force_participation": "participacion_laboral",
    "unemployment_rate": "tasa_desempleo",
    "poverty": "pobreza",
    "limited_english": "ingles_limitado",
    "commute_time": "tiempo_traslado",
    "health_uninsured": "sin_seguro_medico",

    "Ozone": "ozono",
    "PM25": "pm25",
    "N02": "no2"
}

train = train.rename(columns=RENOMBRE_COLUMNAS)
test = test.rename(columns=RENOMBRE_COLUMNAS)
sample_submission = sample_submission.rename(columns=RENOMBRE_COLUMNAS)

ID_COL = "id_paciente"
TARGET = "diagnostico_90_dias"

print("Dimensión del dataset de entrenamiento:", train.shape)
print("Dimensión del dataset de prueba:", test.shape)

display(train.head())
```

```python
total_registros = train.shape[0]
total_variables = train.shape[1]
total_predictoras = total_variables - 2
registros_test = test.shape[0]
duplicados_id = train[ID_COL].duplicated().sum()
columnas_con_faltantes = train.isna().sum().gt(0).sum()
tasa_clase_positiva = train[TARGET].mean()

kpis_fase2 = pd.DataFrame({
    "KPI": [
        "Registros de entrenamiento",
        "Registros de prueba",
        "Total de variables en entrenamiento",
        "Variables predictoras",
        "Variable objetivo",
        "Pacientes duplicados por ID",
        "Columnas con valores faltantes",
        "Tasa de diagnóstico dentro de 90 días"
    ],
    "Resultado": [
        total_registros,
        registros_test,
        total_variables,
        total_predictoras,
        TARGET,
        duplicados_id,
        columnas_con_faltantes,
        f"{tasa_clase_positiva:.2%}"
    ]
})

display(kpis_fase2)
```

```python
plt.figure(figsize=(7, 5))

ax = sns.countplot(
    data=train,
    x=TARGET,
    palette=["#d95f43", "#147d78"]
)

plt.title("Distribución de la variable objetivo", fontsize=14, weight="bold")
plt.xlabel("Diagnóstico dentro de 90 días")
plt.ylabel("Cantidad de pacientes")
plt.xticks([0, 1], ["No", "Sí"])

for container in ax.containers:
    ax.bar_label(container)

plt.tight_layout()
plt.savefig(OUTPUT_DIR / "01_distribucion_variable_objetivo.png", dpi=300)
plt.show()

distribucion_target = (
    train[TARGET]
    .value_counts(normalize=False)
    .rename_axis("Clase")
    .reset_index(name="Cantidad")
)

distribucion_target["Porcentaje"] = (
    distribucion_target["Cantidad"] / distribucion_target["Cantidad"].sum()
)

display(distribucion_target)
```

La variable objetivo muestra una diferencia moderada entre sus clases. Aproximadamente el 62.5% de los casos, corresponde a pacientes diagnosticados dentro de los 90 días, mientras que el otro porcentaje restante pertenece a diagnosticos posteriores a ese periodo. Aunque la diferencia no sea muy marcada, se debe tomar en cuenta un entrenamiento del modelo para evitar inclinaciones hacia la clase con más registros.

```python
variables_predictoras = [c for c in train.columns if c not in [ID_COL, TARGET]]

variables_numericas = train[variables_predictoras].select_dtypes(include=["int64", "float64"]).columns.tolist()
variables_categoricas = [c for c in variables_predictoras if c not in variables_numericas]

resumen_tipos = pd.DataFrame({
    "Tipo de variable": ["Numéricas", "Categóricas"],
    "Cantidad": [len(variables_numericas), len(variables_categoricas)]
})

display(resumen_tipos)

plt.figure(figsize=(6, 4))
sns.barplot(data=resumen_tipos, x="Tipo de variable", y="Cantidad", palette=["#147d78", "#d95f43"])
plt.title("Cantidad de variables por tipo", fontsize=14, weight="bold")
plt.xlabel("")
plt.ylabel("Cantidad")
plt.tight_layout()
plt.savefig(OUTPUT_DIR / "02_tipos_variables.png", dpi=300)
plt.show()
```

```python
faltantes = (
    train.isna()
    .mean()
    .sort_values(ascending=False)
    .reset_index()
)

faltantes.columns = ["Variable", "Proporcion_faltante"]
faltantes["Porcentaje_faltante"] = faltantes["Proporcion_faltante"] * 100

display(faltantes.head(20))

plt.figure(figsize=(10, 6))
sns.barplot(
    data=faltantes.head(20),
    y="Variable",
    x="Porcentaje_faltante",
    color="#147d78"
)

plt.title("Top 20 variables con mayor porcentaje de valores faltantes", fontsize=14, weight="bold")
plt.xlabel("Porcentaje de valores faltantes")
plt.ylabel("Variable")
plt.tight_layout()
plt.savefig(OUTPUT_DIR / "03_valores_faltantes.png", dpi=300)
plt.show()
```

La mayoría de pascientes se ubica entre 50 y 70 años, con mayor pico de concentración cerca de los 60. Aunque hay más caso, la variable edad, por si sola, no toma diferencia entre ambos grupos, por eso debe evaluarse junto con otras variables clínicas.


```python
if "tipo_seguro_pagador" in train.columns:
    tasa_seguro = (
        train
        .assign(tipo_seguro_pagador=train["tipo_seguro_pagador"].fillna("Sin dato"))
        .groupby("tipo_seguro_pagador")[TARGET]
        .agg(["count", "mean"])
        .reset_index()
        .rename(columns={
            "count": "Cantidad",
            "mean": "Tasa_diagnostico_90_dias"
        })
        .sort_values("Tasa_diagnostico_90_dias", ascending=False)
    )

    display(tasa_seguro)

    plt.figure(figsize=(9, 5))
    sns.barplot(
        data=tasa_seguro,
        y="tipo_seguro_pagador",
        x="Tasa_diagnostico_90_dias",
        color="#147d78"
    )

    plt.title("Tasa de diagnóstico dentro de 90 días por tipo de seguro/pagador", fontsize=14, weight="bold")
    plt.xlabel("Tasa de diagnóstico dentro de 90 días")
    plt.ylabel("Tipo de seguro o pagador")
    plt.tight_layout()
    plt.savefig(OUTPUT_DIR / "05_tasa_por_seguro.png", dpi=300)
    plt.show()
```

La tasa de diagnóstico dentro de 90 días cambia según el tipo de seguro. El grupo Sin dato tiene la tasa más alta, aunque debe leerse con cuidado porque puede mezclar casos sin información clara. Entre los seguros registrados, Medicare Advantage presenta una tasa mayor que Medicaid y Commercial, lo que sugiere que la cobertura o el tipo de pagador podría estar relacionado con diferencias en el acceso u oportunidad del diagnóstico.

```python
if "region_geografica" in train.columns:
    tasa_region = (
        train
        .assign(region_geografica=train["region_geografica"].fillna("Sin dato"))
        .groupby("region_geografica")[TARGET]
        .agg(["count", "mean"])
        .reset_index()
        .rename(columns={
            "count": "Cantidad",
            "mean": "Tasa_diagnostico_90_dias"
        })
        .sort_values("Tasa_diagnostico_90_dias", ascending=False)
    )

    display(tasa_region)

    plt.figure(figsize=(9, 5))
    sns.barplot(
        data=tasa_region,
        y="region_geografica",
        x="Tasa_diagnostico_90_dias",
        color="#d95f43"
    )

    plt.title("Tasa de diagnóstico dentro de 90 días por región geográfica", fontsize=14, weight="bold")
    plt.xlabel("Tasa de diagnóstico dentro de 90 días")
    plt.ylabel("Región")
    plt.tight_layout()
    plt.savefig(OUTPUT_DIR / "06_tasa_por_region.png", dpi=300)
    plt.show()
```

La tasa de diagnóstico dentro de 90 días presenta diferencias entre regiones. West y Midwest muestran los valores más altos, mientras que South y Northeast tienen tasas menores. Esto sugiere que la ubicación podría influir en el acceso u oportunidad del diagnóstico.

La categoría Sin dato debe interpretarse con cuidado, ya que tiene pocos registros. En el contexto peruano, este análisis sería útil para estudiar posibles brechas entre Lima y otras regiones del país, especialmente donde el acceso a servicios oncológicos es más limitado.

Las variables socioeconómicas muestran comportamientos bastante similares entre ambos grupos de pacientes. Si bien se aprecian algunas diferencias en ingreso, pobreza, falta de seguro médico y desempleo, estas no son lo suficientemente fuertes como para distinguir claramente a quienes fueron diagnosticadas dentro de 90 días de quienes no.

En el contexto peruano, su análisis es relevante porque factores como ingreso, pobreza y acceso a seguro pueden reflejar brechas en la atención sanitaria. En esta fase se transforman los datos para que puedan ser utilizados por los modelos de Machine Learning.

En el contexto peruano, esta fase es especialmente importante porque los sistemas de salud suelen presentar datos incompletos, heterogéneos o registrados con distintos criterios. Por ello, una preparación adecuada permite reducir errores y mejorar la confiabilidad del análisi

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

X = train.drop(columns=[ID_COL, TARGET])
y = train[TARGET].astype(int)

X_test = test.drop(columns=[ID_COL])

print("X entrenamiento:", X.shape)
print("y:", y.shape)
print("X test:", X_test.shape)
```

```python
variables_categoricas_forzadas = [
    "raza_paciente",
    "tipo_seguro_pagador",
    "estado_paciente",
    "codigo_postal_3",
    "sexo_paciente",
    "codigo_diagnostico_cancer_mama",
    "descripcion_diagnostico_cancer_mama",
    "codigo_diagnostico_cancer_metastasico",
    "primer_tratamiento_novedoso",
    "tipo_primer_tratamiento_novedoso",
    "region_geografica",
    "division_geografica"
]

variables_categoricas_forzadas = [c for c in variables_categoricas_forzadas if c in X.columns]

variables_numericas_modelo = [
    c for c in X.columns
    if c not in variables_categoricas_forzadas and pd.api.types.is_numeric_dtype(X[c])
]

variables_categoricas_modelo = [
    c for c in X.columns
    if c not in variables_numericas_modelo
]

resumen_preparacion = pd.DataFrame({
    "Tipo de variable": ["Numéricas", "Categóricas"],
    "Cantidad": [len(variables_numericas_modelo), len(variables_categoricas_modelo)]
})

display(resumen_preparacion)
```

```python
faltantes_modelo = X.isna().mean().sort_values(ascending=False).reset_index()
faltantes_modelo.columns = ["Variable", "Proporcion_faltante"]
faltantes_modelo["Porcentaje_faltante"] = faltantes_modelo["Proporcion_faltante"] * 100

display(faltantes_modelo.head(20))

umbral_faltantes = 0.60

columnas_muchos_faltantes = (
    faltantes_modelo
    .query("Proporcion_faltante > @umbral_faltantes")["Variable"]
    .tolist()
)

print("Columnas con más de 60% de faltantes:")
print(columnas_muchos_faltantes)
```

Por esta razón, se decidió excluir las variables con más de 60% de valores faltantes, evitando que el modelo aprenda a partir de información incompleta o poco representativa. Las demás variables tienen niveles de ausencia bajos y pueden tratarse mediante imputación.

```python
X = X.drop(columns=columnas_muchos_faltantes)
X_test = X_test.drop(columns=columnas_muchos_faltantes)

variables_numericas_modelo = [c for c in variables_numericas_modelo if c in X.columns]
variables_categoricas_modelo = [c for c in variables_categoricas_modelo if c in X.columns]

print("X después de eliminar columnas con muchos faltantes:", X.shape)
```

```python
columnas_excluir = []

if "descripcion_diagnostico_cancer_mama" in X.columns:
    columnas_excluir.append("descripcion_diagnostico_cancer_mama")

for col in variables_categoricas_modelo:
    if X[col].nunique(dropna=True) > 120:
        columnas_excluir.append(col)

columnas_excluir = sorted(set(columnas_excluir))

print("Columnas excluidas del modelado:")
print(columnas_excluir)

X = X.drop(columns=columnas_excluir)
X_test = X_test.drop(columns=columnas_excluir)

variables_numericas_modelo = [c for c in variables_numericas_modelo if c in X.columns]
variables_categoricas_modelo = [c for c in variables_categoricas_modelo if c in X.columns]

print("X final para modelado:", X.shape)
```

```python
X_train, X_valid, y_train, y_valid = train_test_split(
    X,
    y,
    test_size=0.20,
    stratify=y,
    random_state=42
)

resumen_split = pd.DataFrame({
    "Conjunto": ["Entrenamiento", "Validación"],
    "Registros": [X_train.shape[0], X_valid.shape[0]],
    "Porcentaje clase positiva": [y_train.mean(), y_valid.mean()]
})

display(resumen_split)
```

La división entre entrenamiento y validación mantuvo prácticamente la misma proporción de casos positivos en ambos conjuntos: 62.46% en entrenamiento y 62.43% en validación. Esto indica que la partición fue adecuada y conservó la distribución original de la variable objetivo, evitando sesgos entre los datos usados para entrenar y los usados para evaluar el modelo.

```python
try:
    encoder = OneHotEncoder(handle_unknown="ignore", sparse_output=False)
except TypeError:
    encoder = OneHotEncoder(handle_unknown="ignore", sparse=False)

pipeline_numerico = Pipeline(steps=[
    ("imputacion_mediana", SimpleImputer(strategy="median")),
    ("escalamiento", StandardScaler())
])

pipeline_categorico = Pipeline(steps=[
    ("imputacion_sin_dato", SimpleImputer(strategy="constant", fill_value="Sin dato")),
    ("codificacion_onehot", encoder)
])

preprocesador = ColumnTransformer(
    transformers=[
        ("variables_numericas", pipeline_numerico, variables_numericas_modelo),
        ("variables_categoricas", pipeline_categorico, variables_categoricas_modelo)
    ]
)

preprocesador
```

El "ColumnTransformer" permite aplicar tratamientos distintos según el tipo de variable. Las variables numéricas fueron imputadas y escaladas, mientras que las categóricas fueron imputadas y codificadas con "OneHotEncoder". Este proceso garantiza que la matriz final no tenga valores faltantes y que todas las variables estén en formato numérico para el entrenamiento de los modelos.

```python
X_train_preparado = preprocesador.fit_transform(X_train)
X_valid_preparado = preprocesador.transform(X_valid)

print("Dimensión original de X_train:", X_train.shape)
print("Dimensión transformada de X_train:", X_train_preparado.shape)
print("Dimensión transformada de X_valid:", X_valid_preparado.shape)
```

El incremento de 76 a 232 variables después del preprocesamiento es normal, ya que las variables categóricas fueron convertidas mediante One-Hot Encoding. Con ello, todos los datos quedaron en formato numérico y estandarizado para ser utilizados por los modelos de clasificación.

# Fase 4 CRISP-DM: Modelado

En esta fase se construyen los modelos predictivos utilizando el dataset previamente preparado.  
El objetivo es entrenar diferentes algoritmos de clasificación supervisada y comparar su desempeño inicial mediante validación cruzada.

Los modelos seleccionados son:

1. **K-Nearest Neighbors (KNN)**  
   Modelo basado en distancia. Clasifica una observación según las clases de sus vecinos más cercanos.

2. **Regresión Logística**  
   Modelo lineal interpretable que estima la probabilidad de pertenecer a una clase.

3. **Support Vector Machine (SVM)**  
   Modelo de margen máximo que busca separar las clases mediante una frontera de decisión.

Dado que el problema corresponde a una clasificación binaria, todos los modelos se entrenan para predecir si el diagnóstico de cáncer de mama ocurrió dentro de los primeros 90 días.

En esta etapa se utilizará `GridSearchCV` para buscar combinaciones de hiperparámetros y seleccionar la mejor configuración de cada modelo según el F1-score.

## Criterio principal de evaluación del modelo

Para seleccionar el mejor modelo se utiliza principalmente el F1-score. Esta métrica combina precision y recall, por lo que resulta más adecuada que accuracy cuando existe desbalance entre clases.

En este problema, el interés no es solamente acertar en términos generales, sino identificar correctamente a las pacientes con diagnóstico dentro de 90 días. Por esa razón, también se revisan métricas complementarias como recall, AUC-ROC y Average Precision.

El recall es especialmente importante en salud porque permite medir cuántos casos positivos logra detectar el modelo. Sin embargo, seleccionar únicamente por recall podría generar demasiados falsos positivos. Por ello, el F1-score se utiliza como criterio principal, ya que busca un equilibrio entre detectar pacientes positivas y mantener una proporción razonable de aciertos.

```python
from sklearn.model_selection import GridSearchCV, StratifiedKFold
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC
from sklearn.calibration import CalibratedClassifierCV
from sklearn.feature_selection import SelectKBest, mutual_info_classif
from sklearn.pipeline import Pipeline

from sklearn.metrics import make_scorer, f1_score

import pandas as pd
import numpy as np
```

```python
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

metrica_principal = "f1"
```

## Modelo K-Nearest Neighbors


```python
from sklearn.model_selection import GridSearchCV, StratifiedKFold
from sklearn.neighbors import KNeighborsClassifier
from sklearn.feature_selection import SelectKBest, mutual_info_classif
from sklearn.pipeline import Pipeline

cv_knn = StratifiedKFold(
    n_splits=3,
    shuffle=True,
    random_state=42
)

pipeline_knn = Pipeline(steps=[
    ("preprocesamiento", preprocesador),
    ("seleccion_variables", SelectKBest(score_func=mutual_info_classif, k=40)),
    ("modelo", KNeighborsClassifier())
])

parametros_knn = {
    "modelo__n_neighbors": [15, 31],
    "modelo__weights": ["distance"],
    "modelo__metric": ["manhattan"]
}

grid_knn = GridSearchCV(
    estimator=pipeline_knn,
    param_grid=parametros_knn,
    scoring="f1",
    cv=cv_knn,
    n_jobs=-1,
    verbose=1
)

grid_knn.fit(X_train, y_train)

print("Mejores hiperparámetros KNN:")
print(grid_knn.best_params_)

print("Mejor F1 promedio en validación cruzada:")
print(grid_knn.best_score_)
```

## Modelo Regresión Logística


```python
pipeline_logistica = Pipeline(steps=[
    ("preprocesamiento", preprocesador),
    ("modelo", LogisticRegression(
        max_iter=2000,
        class_weight="balanced",
        random_state=42
    ))
])

parametros_logistica = {
    "modelo__C": [0.01, 0.1, 1, 3, 10],
    "modelo__penalty": ["l2"],
    "modelo__solver": ["lbfgs", "liblinear"]
}

grid_logistica = GridSearchCV(
    estimator=pipeline_logistica,
    param_grid=parametros_logistica,
    scoring=metrica_principal,
    cv=cv,
    n_jobs=-1,
    verbose=1
)

grid_logistica.fit(X_train, y_train)

print("Mejores hiperparámetros Regresión Logística:")
print(grid_logistica.best_params_)

print("Mejor F1 promedio en validación cruzada:")
print(grid_logistica.best_score_)
```

## Modelo Support Vector Machine


```python
pipeline_svm = Pipeline(steps=[
    ("preprocesamiento", preprocesador),
    ("modelo", CalibratedClassifierCV(
        estimator=LinearSVC(
            class_weight="balanced",
            random_state=42,
            max_iter=5000
        ),
        cv=3,
        method="sigmoid"
    ))
])

parametros_svm = {
    "modelo__estimator__C": [0.01, 0.1, 1, 3, 10]
}

grid_svm = GridSearchCV(
    estimator=pipeline_svm,
    param_grid=parametros_svm,
    scoring=metrica_principal,
    cv=cv,
    n_jobs=-1,
    verbose=1
)

grid_svm.fit(X_train, y_train)

print("Mejores hiperparámetros SVM:")
print(grid_svm.best_params_)

print("Mejor F1 promedio en validación cruzada:")
print(grid_svm.best_score_)
```

# Fase 5 CRISP-DM: Evaluación de modelos

En esta fase se evalúan los modelos entrenados utilizando el conjunto de validación separado previamente.  
A diferencia de la fase de modelado, donde se utilizó validación cruzada para seleccionar hiperparámetros, en esta etapa se mide el desempeño final de cada modelo sobre datos no utilizados directamente en el entrenamiento.

Los modelos evaluados son:

- K-Nearest Neighbors
- Regresión Logística
- Support Vector Machine

La evaluación se realizará mediante métricas de clasificación, matrices de confusión, curvas ROC, curvas Precision-Recall y pruebas estadísticas. Debido al desbalance moderado observado en la variable objetivo, no se utilizará únicamente accuracy, sino también precision, recall, F1-score y AUC-ROC.

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix,
    classification_report,
    RocCurveDisplay,
    PrecisionRecallDisplay
)

import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

try:
    modelos_entrenados = mejores_modelos
except NameError:
    modelos_entrenados = {
        "KNN": grid_knn.best_estimator_,
        "Regresión Logística": grid_logistica.best_estimator_,
        "SVM": grid_svm.best_estimator_
    }

modelos_entrenados
```

```python
def obtener_probabilidades(modelo, X):

    if hasattr(modelo, "predict_proba"):
        return modelo.predict_proba(X)[:, 1]

    if hasattr(modelo, "decision_function"):
        scores = modelo.decision_function(X)
        return 1 / (1 + np.exp(-scores))

    return modelo.predict(X)
```

```python
resultados_evaluacion = []
predicciones_modelos = {}
probabilidades_modelos = {}

for nombre_modelo, modelo in modelos_entrenados.items():
    probabilidades = obtener_probabilidades(modelo, X_valid)
    predicciones = (probabilidades >= 0.5).astype(int)

    predicciones_modelos[nombre_modelo] = predicciones
    probabilidades_modelos[nombre_modelo] = probabilidades

    tn, fp, fn, tp = confusion_matrix(y_valid, predicciones).ravel()

    resultados_evaluacion.append({
        "Modelo": nombre_modelo,
        "Accuracy": accuracy_score(y_valid, predicciones),
        "Precision": precision_score(y_valid, predicciones, zero_division=0),
        "Recall": recall_score(y_valid, predicciones, zero_division=0),
        "F1-score": f1_score(y_valid, predicciones, zero_division=0),
        "AUC-ROC": roc_auc_score(y_valid, probabilidades),
        "Average Precision": average_precision_score(y_valid, probabilidades),
        "Verdaderos negativos": tn,
        "Falsos positivos": fp,
        "Falsos negativos": fn,
        "Verdaderos positivos": tp
    })

metricas_finales = (
    pd.DataFrame(resultados_evaluacion)
    .sort_values("F1-score", ascending=False)
    .reset_index(drop=True)
)

display(metricas_finales)
```

```python
metricas_grafico = metricas_finales.melt(
    id_vars="Modelo",
    value_vars=["Accuracy", "Precision", "Recall", "F1-score", "AUC-ROC", "Average Precision"],
    var_name="Métrica",
    value_name="Valor"
)

plt.figure(figsize=(11, 6))
sns.barplot(
    data=metricas_grafico,
    x="Métrica",
    y="Valor",
    hue="Modelo"
)

plt.title("Comparación final de métricas por modelo", fontsize=14, weight="bold")
plt.ylim(0, 1)
plt.xlabel("Métrica")
plt.ylabel("Valor")
plt.legend(title="Modelo", loc="lower right")
plt.tight_layout()
plt.savefig(OUTPUT_DIR / "10_comparacion_metricas_finales.png", dpi=300)
plt.show()
```

El gráfico muestra que el modelo SVM tiene el mejor desempeño general, sobre todo en recall y F1-score, que son métricas importantes cuando se busca detectar correctamente los casos positivos. La Regresión Logística también tuvo resultados muy cercanos, especialmente en AUC-ROC y Average Precision, por lo que sigue siendo una alternativa competitiva. KNN quedó por debajo en casi todas las métricas, lo que indica que se adaptó peor a este conjunto de datos.

```python
def bootstrap_f1(y_real, y_predicha, n_iteraciones=1000, random_state=42):
    rng = np.random.default_rng(random_state)
    y_real = np.array(y_real)
    y_predicha = np.array(y_predicha)

    resultados = []

    for _ in range(n_iteraciones):
        indices = rng.integers(0, len(y_real), len(y_real))
        f1 = f1_score(
            y_real[indices],
            y_predicha[indices],
            zero_division=0
        )
        resultados.append(f1)

    return np.percentile(resultados, [2.5, 50, 97.5])


bootstrap_resultados = []

for nombre_modelo, predicciones in predicciones_modelos.items():
    limite_inferior, mediana, limite_superior = bootstrap_f1(y_valid, predicciones)

    bootstrap_resultados.append({
        "Modelo": nombre_modelo,
        "F1 mediana bootstrap": mediana,
        "Límite inferior IC 95%": limite_inferior,
        "Límite superior IC 95%": limite_superior
    })

bootstrap_df = (
    pd.DataFrame(bootstrap_resultados)
    .sort_values("F1 mediana bootstrap", ascending=False)
    .reset_index(drop=True)
)

display(bootstrap_df)
```

En general, los intervalos son relativamente estrechos, lo que indica que los resultados son estables. Además, el intervalo de SVM se ubica por encima del de KNN y muestra ventaja frente a Regresión Logística, reforzando su selección como modelo final.

```python
from scipy.stats import chi2

def prueba_mcnemar(y_real, pred_a, pred_b):

    y_real = np.array(y_real)
    pred_a = np.array(pred_a)
    pred_b = np.array(pred_b)

    correcto_a = pred_a == y_real
    correcto_b = pred_b == y_real

    b = np.sum(correcto_a & ~correcto_b)
    c = np.sum(~correcto_a & correcto_b)

    if b + c == 0:
        estadistico = 0
        p_valor = 1
    else:
        estadistico = ((abs(b - c) - 1) ** 2) / (b + c)
        p_valor = 1 - chi2.cdf(estadistico, df=1)

    return b, c, estadistico, p_valor


comparaciones_mcnemar = []
nombres_modelos = list(predicciones_modelos.keys())

for i in range(len(nombres_modelos)):
    for j in range(i + 1, len(nombres_modelos)):
        modelo_a = nombres_modelos[i]
        modelo_b = nombres_modelos[j]

        b, c, estadistico, p_valor = prueba_mcnemar(
            y_valid,
            predicciones_modelos[modelo_a],
            predicciones_modelos[modelo_b]
        )

        comparaciones_mcnemar.append({
            "Comparación": f"{modelo_a} vs {modelo_b}",
            "b": b,
            "c": c,
            "Chi-cuadrado": estadistico,
            "p-valor": p_valor,
            "Diferencia significativa al 5%": p_valor < 0.05
        })

mcnemar_df = pd.DataFrame(comparaciones_mcnemar)
display(mcnemar_df)
```

```python
import gradio as gr
import pandas as pd

def etiqueta_variable(nombre):
    return nombre.replace("_", " ").capitalize()


def opciones_variable(columna, max_opciones=25):
    valores = (
        X[columna]
        .fillna("Sin dato")
        .astype(str)
        .value_counts()
        .head(max_opciones)
        .index
        .tolist()
    )
    return valores if len(valores) > 0 else ["Sin dato"]


componentes_entrada = []

for col in variables_demo:
    if col in variables_numericas_modelo:
        componentes_entrada.append(
            gr.Number(
                value=float(valores_defecto[col]),
                label=etiqueta_variable(col)
            )
        )
    else:
        componentes_entrada.append(
            gr.Dropdown(
                choices=opciones_variable(col),
                value=str(valores_defecto[col]),
                label=etiqueta_variable(col)
            )
        )


def predecir_diagnostico(*valores):
    fila = valores_defecto.copy()

    for columna, valor in zip(variables_demo, valores):
        fila[columna] = valor

    paciente = pd.DataFrame([fila])[X.columns]

    probabilidad = float(obtener_probabilidades(modelo_final_entrenado, paciente)[0])
    clase_predicha = int(probabilidad >= 0.5)

    if probabilidad >= 0.67:
        interpretacion = "Probabilidad alta de diagnóstico dentro de 90 días"
    elif probabilidad >= 0.45:
        interpretacion = "Probabilidad intermedia de diagnóstico dentro de 90 días"
    else:
        interpretacion = "Probabilidad baja de diagnóstico dentro de 90 días"

    return {
        "Modelo utilizado": modelo_final_nombre,
        "Probabilidad estimada": round(probabilidad, 4),
        "Porcentaje estimado": f"{probabilidad:.2%}",
        "Clase predicha": "Sí" if clase_predicha == 1 else "No",
        "Interpretación": interpretacion,
        "Nota": "Demo académica. No reemplaza evaluación médica."
    }


demo = gr.Interface(
    fn=predecir_diagnostico,
    inputs=componentes_entrada,
    outputs=gr.JSON(label="Resultado de la predicción"),
    title="Demo predictiva: diagnóstico de cáncer de mama dentro de 90 días",
    description=(
        "Ingrese variables clínicas, demográficas, socioeconómicas y ambientales. "
        "La demo estima la probabilidad de diagnóstico dentro de los primeros 90 días. "
        "Uso académico, no clínico."
    ),
    theme=gr.themes.Soft()
)

demo.launch(share=True)
```

## Consideraciones éticas de la implementación

La demo desarrollada tiene fines académicos y demostrativos.  
No debe utilizarse como herramienta clínica real ni como sustituto del juicio médico.

En una implementación real dentro del sistema de salud peruano, se deberían considerar los siguientes aspectos:

- Validación con datos locales de pacientes peruanas.
- Evaluación de sesgos por región, nivel socioeconómico, tipo de seguro y grupo poblacional.
- Revisión por especialistas clínicos.
- Cumplimiento de normas de protección de datos personales.
- Monitoreo del desempeño del modelo en el tiempo.
- Integración controlada con sistemas hospitalarios o institucionales.

Además, debe considerarse que variables como raza, región, tipo de seguro y nivel socioeconómico pueden capturar desigualdades estructurales en el acceso al diagnóstico. Por ello, su uso no debe interpretarse como una justificación para decisiones clínicas automáticas, sino como una oportunidad para auditar brechas de atención. En una implementación real, sería necesario evaluar el desempeño del modelo por subgrupos y verificar que no reproduzca errores sistemáticos contra poblaciones específicas.
